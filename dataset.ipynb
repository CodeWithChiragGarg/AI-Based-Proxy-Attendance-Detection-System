```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

# Defining missing variables to make the cell runnable
students = [f"STU{i:04d}" for i in range(1, 1501)] # Example student IDs
subjects = ["Math", "Science", "History", "Physics", "Chemistry", "Biology", "English", "CS", "ML", "Python", "DBMS", "CN"] # Example subjects
start_date = datetime(2026, 4, 1) # Example start date
end_date = datetime(2026, 4, 30) # Example end date
days = (end_date - start_date).days + 1 # Number of days

# Initialize df if it doesn't exist. This cell assumes 'df' is already populated
# with genuine records, but for demonstration, we create an empty one if not found.
if 'df' not in locals() and 'df' not in globals():
    df = pd.DataFrame(columns=[
        "Student_ID", "Date", "Time", "Subject", "Location", "Device_ID",
        "Latitude", "Longitude", "Attendance_Status", "Actual_Label"
    ])

proxy_records = []

for i in range(600):
    random_student = random.choice(students)
    subject = random.choice(subjects)
    date = start_date + timedelta(days=random.randint(0, days-1))

    suspicious_time = datetime(date.year, date.month, date.day, random.choice([9,10,11,12,14,15]), random.randint(0, 2))

    # Same device fraud
    shared_device = f"DEV_PROXY_{random.randint(1, 20)}"

    # Location mismatch
    wrong_lat = 30.5000 + np.random.normal(0, 0.001)
    wrong_lon = 78.0000 + np.random.normal(0, 0.001)

    proxy_records.append({
        "Student_ID": random_student,
        "Date": date.strftime("%Y-%m-%d"),
        "Time": suspicious_time.strftime("%H:%M:%S"),
        "Subject": subject,
        "Location": f"Room_{subject}",
        "Device_ID": shared_device,
        "Latitude": wrong_lat,
        "Longitude": wrong_lon,
        "Attendance_Status": "Present",
        "Actual_Label": "Proxy"
    })

proxy_df = pd.DataFrame(proxy_records)

df = pd.concat([df, proxy_df], ignore_index=True)

df = df.sample(frac=1, random_state=42).reset_index(drop=True)

df.head()
```
